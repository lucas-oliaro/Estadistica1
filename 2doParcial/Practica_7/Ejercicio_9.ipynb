Simular un conjunto de datos de tamaño 𝑛 = 10 a partir de una distribución normal con 𝜃 = 5
y 𝜎 = 1. Graficar la likelihood. Usar el prior 𝜃 ∼ N (0, 0.5) y calcular la distribución posterior de
𝜃. Graficar las distribuciones prior y posterior.


```python
from scipy.stats import norm

n_muestras = 10
# Uso norm para generar datos y tomo muestras aleatorias con rvs
datos = norm.rvs(loc=5, scale=1, size=n_muestras) # loc = media, scale = desviacion estandar
```

```python
from matplotlib import pyplot as plt
import numpy as np

theta = np.linspace(-3, 10, 10)
likelihood = np.prod([norm.pdf(datos, loc=th, scale=1) for th in theta], axis=1)

plt.plot(theta, likelihood, label="Likelihood")
plt.xlabel("theta")
plt.ylabel("Likelihood")
plt.title("Likelihood de theta dados los datos")
plt.legend()
plt.show()
```

Ahora uso el prior propuesto para calcular la posterior de 𝜃:   

```python
prior = norm.pdf(theta, loc=0, scale=0.5)

#likelihood = np.prod([norm.pdf(datos, loc=th, scale=1) for th in theta], axis=1)
posterior = likelihood * prior
posterior /= posterior.sum()  
```

```python
plt.plot(theta, prior, label="Prior")
plt.plot(theta, posterior, label="Posterior")
plt.xlabel("Theta")
plt.ylabel("Densidad")
plt.title("Prior y Posterior")
plt.legend()
plt.show()
```
